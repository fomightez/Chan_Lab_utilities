```python
from IPython.display import HTML, display
def build_table_html(title, df, header_color):
    """Helper to convert a DataFrame into a styled HTML card without a column header row."""
    rows = ""
    for _, row in df.iterrows():
        rows += f"""
        <tr>
            <td class="container-cell">{row.iloc[0]}</td>
            <td class="z-cell">{row.iloc[1]}</td>
        </tr>
        """

    return f"""
    <div class="table-card">
        <div class="card-header" style="background-color: {header_color};">
            {title}
        </div>
        <table class="data-table">
            <tbody>
                {rows}
            </tbody>
        </table>
    </div>
    """

# Build the individual table cards with pastels similar to your Excel style
html_flasks = build_table_html("Flasks", df_flasks, "#D9E1F2")  # Muted Blue-Gray
html_dishes = build_table_html("Dishes", df_dishes, "#E2EFDA")  # Soft Sage
html_wellplates = build_table_html(
    "Well Plates", df_wellplates, "#FFF2CC"
)  # Warm pastel

# --- 2. ASSEMBLE FULL PAGE HTML ---
full_html = f"""
<style>
    /* Styling for poster wall display */
    .lab-guide-container {{
        font-family: 'Segoe UI', Arial, sans-serif;
        max-width: 800px;
        margin: 0 auto;
        padding: 20px;
        color: #000;
        background-color: #ffffff;
    }}
    .guide-title {{
        text-align: center;
        font-size: 38px;
        font-weight: 800;
        letter-spacing: 1px;
        padding: 12px;
        background-color: #FCE4D6; /* Muted Coral */ 
        border: 2px solid #D9D9D9;
        border-radius: 8px;
        margin-bottom: 24px;
    }}
    .tables-grid {{
        display: grid;
        grid-template-columns: 1fr 1fr;
        gap: 20px;
    }}
    .table-card {{
        border: 2px solid #BFBFBF;
        border-radius: 8px;
        overflow: hidden;
        box-shadow: 0 2px 5px rgba(0,0,0,0.05);
    }}
    .card-header {{
        font-size: 24px;
        font-weight: bold;
        text-align: center;
        padding: 8px 0;
        border-bottom: 2px solid #BFBFBF;
    }}
    .data-table {{
        width: 100%;
        border-collapse: collapse;
    }}
    .data-table th {{
        font-size: 20px;
        background-color: #F2F2F2;
        padding: 8px;
        border-bottom: 2px solid #D9D9D9;
    }}
    .data-table td {{
        font-size: 22px;
        padding: 10px 14px;
        border-bottom: 1px solid #E0E0E0;
    }}
    .container-cell {{
        font-weight: bold;
        text-align: left;
    }}
    .z-cell {{
        text-align: right;
        font-family: 'Courier New', monospace;
        font-weight: 700;
    }}
    .guide-footer {{
        margin-top: 25px;
        text-align: center;
        font-size: 18px;
        font-style: italic;
        color: #444;
        border-top: 1px dashed #CCC;
        padding-top: 12px;
    }}
    
    /* Print optimization rules */
    @media print {{
        body * {{
            visibility: hidden;
        }}
        .lab-guide-container, .lab-guide-container * {{
            visibility: visible;
        }}
        .lab-guide-container {{
            position: absolute;
            left: 0;
            top: 0;
            width: 100%;
        }}
    }}
</style>

<div class="lab-guide-container">
    <div class="guide-title">ZOE FOCUS / Z-POSITION GUIDE</div>
    
    <div class="tables-grid">
        {html_flasks}
        <div style="display: flex; flex-direction: column; gap: 20px;">
            {html_dishes}
            {html_wellplates}
        </div>
    </div>
    
    <div class="guide-footer">
        Please inform Wayne if you have values to fill in or new container types to add.
    </div>
</div>
"""

display(HTML(full_html))

# Save the rendered HTML to a file on disk
with open("ZOE_Lab_Wall_Guide.html", "w", encoding="utf-8") as f:
    f.write(full_html)

print("Saved to ZOE_Lab_Wall_Guide.html!")
```
